```python
Genero = ['M','F','M']
```

```python
# La llave
ID = [1001,1002,1003]
```

```python
BaseDatos = {}

for i in range(n):
  # La base conecta un diccionario con otro diccionario
  BaseDatos[ID[i]] = {
      'Nombre': Nombres[i],
      'Apellido': Apellidos[i],
      'Edad': Edad[i],
      'Genero': Genero[i]
  }
```

```python
# La visualizacion
import seaborn as sns
```

```python
edades = []
for i in range(n):
  edades.append(BaseDatos[ID[i]]['Edad'])
edades
```

```python
# Como sacamos la informacion de la base de datos
# persona es el iterador
edades = [ persona['Edad'] for persona in BaseDatos.values()]
edades
```

```python
nombres = [ persona['Nombre'] for persona in BaseDatos.values() ]
nombres
```

```python
# ACA DIBUJAMOS
sns.catplot(x=nombres, y=edades, kind='bar')
```

```python
# Hagamos un primer filtro donde escogemos a los hombre
# La nocion de filtrado o mascara
edades = [ persona['Edad'] for persona in BaseDatos.values() if persona['Genero'] == 'M' ]
edades
```

```python
nombres = [ persona['Nombre'] for persona in BaseDatos.values() if persona['Genero'] == 'M' ]
nombres
```

```python
# ACA DIBUJAMOS
sns.catplot(x=nombres, y=edades, kind='bar')
```
